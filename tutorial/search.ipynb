# Search Introduction

*Search* in Text-Fabric is a template based way of looking for structural patterns in your dataset.

Within Text-Fabric we have the unique possibility to combine the ease of formulating search templates for
complicated patterns with the power of programmatically processing the results.

This notebook will show you how to get up and running.

# Incantation

The ins and outs of installing Text-Fabric, getting the corpus, and initializing a notebook are
explained in the [start tutorial](start.ipynb).

```python
from tf.app import use
```

```python
A = use("rdtaylorjr/hexapla", hoist=globals())
```

# Basic search command

We start with the most simple form of issuing a query.
Let's search for the readings of Aquila that the Mercati palimpsest, Rahlfs 1098, attests,
and show the words in them.

```python
query = """
text text=aquila
  word_group siglum~1098
    word
"""
results = A.search(query)
```

We have the results. We only need to display them. Here they are in a table:

```python
A.table(results, end=5)
```

Here is the first one in a pretty display:

```python
A.show(results, end=1)
```

# Condense results

There are two fundamentally different ways of presenting the results: condensed and uncondensed.

In **uncondensed** view, all results are listed individually.
In **condensed** view all nodes of all results are grouped in containers first (e.g. verses), and then presented
container by container.

```python
A.table(results, end=5, condensed=True)
```

# Custom highlighting

We can apply different highlight colours to different parts of the result.

```python
query = """
text text=secunda
  word_group reading>0
    word
"""

results = A.search(query)
```

```python
A.table(results, end=5, condensed=False, colorMap={2: "cyan", 3: "magenta"})
```

# Constraining order

You can stipulate an order on the parts of your template.
Here we look for a stichos of Aquila in which a reading attested by 1098 precedes another one.

```python
query = """
text text=aquila
  stichos
    word_group siglum~1098
    < word_group siglum~1098
"""

results = A.search(query)
```

```python
A.table(results, end=5, colorMap={3: "cyan", 4: "magenta"})
```

# Languages

The feature `languageISO` of a word gives the language and script of its column.
The Secunda is Hebrew in Greek letters.

```python
query = """
word languageISO=heb-Grek
"""

results = A.search(query)
```

```python
A.table(results, end=5)
```
